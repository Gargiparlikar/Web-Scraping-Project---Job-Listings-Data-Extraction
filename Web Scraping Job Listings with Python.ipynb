# Web Scraping Job Listings with Python

This project focuses on extracting job listings from a website using Web Scraping techniques in Python. By utilizing the BeautifulSoup library, job-related details such as the job title, company name, location, and application link are collected from the webpage. The extracted data is then organized into a structured format using Pandas, allowing easy analysis, filtering, and storage.

```python

import requests
from bs4 import BeautifulSoup



```

```python
URL = "https://realpython.github.io/fake-jobs/"
page = requests.get(URL)


```

```python
# Parse the HTML content using BeautifulSoup
soup = BeautifulSoup(page.content, "html.parser")

# Locate the main container that holds all job postings
results = soup.find(id="ResultsContainer")

```

```python
# Find all job posts that contain the word 'python' in the title
python_jobs = results.find_all(
    "h2", string=lambda text: "python" in text.lower()
)

```

```python
# Climb up the HTML structure to get the full card for each job post
python_job_cards = [
    h2_element.parent.parent.parent for h2_element in python_jobs
]

jobs_list = []  # Store job info

for job_card in python_job_cards:
    title_element = job_card.find("h2", class_="title")
    company_element = job_card.find("h3", class_="company")
    location_element = job_card.find("p", class_="location")
    link_url = job_card.find_all("a")[1]["href"]

    # ✅ Print each job detail
    print("Title:", title_element.text.strip())
    print("Company:", company_element.text.strip())
    print("Location:", location_element.text.strip())
    print("Apply Link:", link_url)
    print("-" * 50)  # Separator for readability

    # ✅ Store data into list
    jobs_list.append([
        title_element.text.strip(),
        company_element.text.strip(),
        location_element.text.strip(),
        link_url
    ])
```

```python
# ✅ Save to CSV
import csv
with open("python_jobs.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["Title", "Company", "Location", "Apply Link"])
    writer.writerows(jobs_list)   # <-- FIX HERE

print("✅ CSV File Created: python_jobs.csv")
```
